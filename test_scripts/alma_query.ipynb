```python
import json
import pyvo
import astropy.coordinates
import astropy.units as u
import numpy as np

from astroquery.alma import Alma
from astropy.time import Time
from collections import defaultdict
```

```python
alma = Alma()   
```

```python
# Resolve position
coord = astropy.coordinates.SkyCoord.from_name("HH 212")

BAND_FILTER = ["6", "7"]
SEARCH_RADIUS_ARCSEC = 60
MAX_RESOLUTION_ARCSEC = 1

tbl = alma.query_region(
    coord,
    radius=SEARCH_RADIUS_ARCSEC * u.arcsec,
    band_list=BAND_FILTER,
    public=True,
    science=True,
)

tbl = tbl[tbl["spatial_resolution"] < MAX_RESOLUTION_ARCSEC] # Note: It looks like the spatial_resolution column is in arcseconds / is actually the anngular resolution
# tbl = tbl[tbl["dataproduct_type"] == "image"]   # continuum only, exclude cubes

observations = []
for row in tbl:
    bands = sorted({int(b) for b in str(row["band_list"]).split()})
    observations.append({
        "project_code": row["proposal_id"],
        "alma_source_name": row["target_name"],
        "ra_deg": float(row["s_ra"]),
        "dec_deg": float(row["s_dec"]),
        "band": bands,
        "angular_resolution_arcsec": float(row["spatial_resolution"]),
        "qa2_status": str(row["qa2_passed"]),
    })

all_bands = sorted({b for obs in observations for b in obs["band"]})
resolutions = [obs["angular_resolution_arcsec"] for obs in observations]

report = {
    "target_name": "HH 212",
    "resolved_ra_deg": coord.ra.deg,
    "resolved_dec_deg": coord.dec.deg,
    "search_radius_arcsec": SEARCH_RADIUS_ARCSEC,
    "band_filter": [6, 7],
    "max_angular_resolution_arcsec": MAX_RESOLUTION_ARCSEC,
    "num_observations": len(observations),
    "observations": observations,
    "summary": (
        f"HH 212 has {len(observations)} public ALMA Band 6/7 science "
        f"observations with sub-arcsecond resolution, covering bands "
        f"{all_bands}, with angular resolutions ranging from "
        f"{min(resolutions):.3f} to {max(resolutions):.3f} arcsec across "
        f"{len(np.unique([o['project_code'] for o in observations]))} distinct proposals."
    ) if observations else (
        "No public ALMA Band 6/7 science observations with sub-arcsecond "
        "resolution were found for HH 212."
    ),
}

# with open("/app/output.json", "w") as f:
#     json.dump(report, f, indent=2)
```
